### TASK: Perform K-means clustering algorithm on the user profile feature vectors


With the user profile dataset ready, you need to use the `KMeans` class provided by scikit-learn library to perform clustering on the user profile feature vectors.


For `KMeans` algorithm, one important hyperparameter is the number of clusters `n_cluster`, and a good way to find the optimized `n_cluster` is using to grid search a list of candidates and find the one with the best or optimized clustering evaluation metrics such as minimal `sum of squared distance`:


_TODO: grid search the optimized n_cluster for KMeans() model_


````python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Lista de posibles valores para n_clusters
list_k = list(range(1, 30))  # puedes ajustar el rango si quieres
inertia_list = []

# Iteramos sobre cada k y calculamos la suma de cuadrados dentro del cluster (inertia)
for k in list_k:
    kmeans = KMeans(n_clusters=k, random_state=rs, n_init=10)
    kmeans.fit(features)
    inertia_list.append(kmeans.inertia_)

# Graficamos el método del codo
plt.figure(figsize=(10,6))
plt.plot(list_k, inertia_list, marker='o')
plt.xlabel('Número de clusters (k)')
plt.ylabel('Inertia (suma de cuadrados dentro del cluster)')
plt.title('Método del codo para determinar el número óptimo de clusters')
plt.xticks(list_k)
plt.grid(True)
plt.show()
````

The cluster labels you generated is a list of integers indicating cluster indices. You may use the following utility method to combine the cluster labels and user ids to a dataframe, so that you know which cluster a user belongs:


````python
import pandas as pd

# Suponiendo que user_ids es un DataFrame con una columna 'user_id'
# y cluster_labels es el array generado por KMeans
def combine_cluster_labels(user_ids, labels):
    # Convertir labels a DataFrame
    labels_df = pd.DataFrame(labels)

    # Combinar user_ids con labels usando el índice
    cluster_df = pd.merge(user_ids, labels_df, left_index=True, right_index=True)

    # Renombrar columnas
    cluster_df.columns = ['user', 'cluster']
    return cluster_df

# Aplicar la función
user_clusters = combine_cluster_labels(user_ids, cluster_labels)

# Mostrar los primeros usuarios con su cluster
user_clusters.head()
````

If we plot a covariance matrix of the user profile feature vectors with 14 features, we can observe that some features are actually correlated:


````python
sns.set_theme(style="white")

# Compute the correlation matrix
corr = features.cov()

# Generate a mask for the upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool))

# Set up the matplotlib figure
f, ax = plt.subplots(figsize=(11, 9))

# Generate a custom diverging colormap
cmap = sns.diverging_palette(230, 20, as_cmap=True)

# Draw the heatmap with the mask and correct aspect ratio
sns.heatmap(corr, mask=mask, cmap=cmap, vmax=.3, center=0,
            square=True, linewidths=.5, cbar_kws={"shrink": .5})


plt.show()
````

For example, the feature `MachineLearning` and the feature `DataScience` are correlated. Such covariances among features may indicate that we can apply PCA to find its main components (eigenvectors with max eigenvalues on the covariance matrix).

If we only keep the independent main components, then we can reduce the dimensions of our user profile feature vectors.


Now let's apply the `PCA()` provided by  `scikit-learn` to find the main components in user profile feature vectors and see if we can reduce its dimensions by only keeping the main components.


Note that when calling the  `PCA()` class, there is also an import argument called `n_components` which indicates how many components you want to keep in the PCA result. One way to find an optimized `n_components` is to do a grid search on a list of argument candidates (such as `range(1, 15)`) and calculate the ratio of the accumulated variance for each candidate.

If the accumulated variances ratio of a candidate `n_components` is larger than a threshold, e.g., 90%, then we can say the transformed `n_components` could explain about 90% of variances of the original data variance and can be considered as an optimized components size.


_TODO: Find the optimized `n_components` for PCA_


````python
from sklearn.decomposition import PCA
import numpy as np

# Extraemos las features de los usuarios (sin la columna 'user')
features = user_profile_df.loc[:, user_profile_df.columns != 'user']

# Lista para almacenar la varianza acumulada para cada número de componentes
accumulated_variances = []

# Probamos desde 1 hasta 14 componentes
for n in range(1, features.shape[1] + 1):
    pca = PCA(n_components=n)
    pca.fit(features)
    accumulated_variance = sum(pca.explained_variance_ratio_)
    accumulated_variances.append(accumulated_variance)
    print(f"n_components={n}, accumulated variance={accumulated_variance:.4f}")

# Encontramos el menor n_components que explique al menos 90% de la varianza
threshold = 0.9
optimized_n_components = next(n for n, var in enumerate(accumulated_variances, start=1) if var >= threshold)

print(f"\nOptimized number of PCA components to explain 90% variance: {optimized_n_components}")
````

Once you found the optimized `n_component` argument value, you can apply PCA on the user profile feature vectors and reduce the  14 features into  `n_component` features.


_TODO: Perform PCA to transform original user profile features_


````python
from sklearn.decomposition import PCA
import pandas as pd

# Lista de candidatos para n_components
candidate_components = range(1, 15)
threshold_variance = 0.9  # 90% de varianza acumulada

# Buscamos el número mínimo de componentes que supere el umbral
for n in candidate_components:
    pca = PCA(n_components=n)
    transformed_features = pca.fit_transform(features)
    accumulated_variance = sum(pca.explained_variance_ratio_)
    print(f"n_components={n}, accumulated variance={accumulated_variance:.4f}")
    if accumulated_variance >= threshold_variance:
        optimized_n_components = n
        break

# Aplicamos PCA con los componentes optimizados
pca = PCA(n_components=optimized_n_components)
pca_transformed = pca.fit_transform(features)

# Convertimos a DataFrame y agregamos user_ids
pca_df = pd.DataFrame(pca_transformed, columns=[f'PC{i+1}' for i in range(optimized_n_components)])
pca_df = pd.merge(user_ids, pca_df, left_index=True, right_index=True)

# Mostramos las primeras filas
pca_df.head()
````

### TASK: Generate course recommendations based on the popular courses in the same cluster


The Intuition of clustering-based course recommendation is very simple and can be illustrated via the following example:


Suppose a user has joined a machine learning group (via clustering algorithm). In the group, he/she finds that the top-3 courses enrolled by all other group members are `Machine Learning for Everyone`, `Machine Learning with Python`, `Machine Learning with Scikit-learn`. Since the user has already completed the `Machine Learning for Everyone` earlier, he/she decides to trust the group members' choices and enroll in other two unselected courses `Machine Learning with Python` and `Machine Learning with Scikit-learn`.


In summary, the clustering-based recommender system first groups all users based on their profiles, and maintains a popular courses list for each group.

For any group member who needs course recommendations, the algorithm recommends the unselected courses from the popular course lists.


Next, suppose we have a set of test users, and we want to recommend new courses to them using a clustering-based recommender system:


````python
test_user_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMSkillsNetwork-ML0321EN-Coursera/labs/v2/module_3/ratings.csv"

# Reading the test user data CSV file into a pandas DataFrame
# Selecting only the 'user' and 'item' columns using indexing
test_users_df = pd.read_csv(test_user_url)[['user', 'item']]

# Displaying the first few rows of the DataFrame to inspect its contents
test_users_df.head()
````

The test users dataset has only two columns, the user id and the enrolled course id.


For each user, let's find its cluster label using the k-means results you have performed in previous steps, assuming it is named `cluster_df`.


You can assign the cluster label to all test users via merging the clustering labels (`cluster_df`:):


````python
# Asumiendo que pca_cluster_df tiene columnas: 'user' y 'cluster'
# test_users_df tiene columnas: 'user' y 'item'

# Merge para asignar cluster a cada test user
test_users_labelled = pd.merge(test_users_df, pca_cluster_df, on='user', how='left')

# Mostramos las primeras filas
test_users_labelled.head()
````

The merged the test dataset may look like the following:


![](https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-ML321EN-SkillsNetwork/labs/module_3/images/test_users_label.png)


From the above dataframe, we know each user's enrolled courses and its cluster index.


If we use a `groupby`  and `sum` aggregation, we can get the enrollments count for each course in each group, like the following code snippet:


````python
# Extracting the 'item' and 'cluster' columns from the test_users_labelled DataFrame
courses_cluster = test_users_labelled[['item', 'cluster']]

# Adding a new column 'count' with a value of 1 for each row in the courses_cluster DataFrame
courses_cluster['count'] = [1] * len(courses_cluster)

# Grouping the DataFrame by 'cluster' and 'item', aggregating the 'count' column with the sum function,
# and resetting the index to make the result more readable
courses_cluster_grouped = courses_cluster.groupby(['cluster','item']).agg(enrollments=('count','sum')).reset_index()
````

_TODO: For each test user, try to recommend any unseen courses based on the popular courses in his/her cluster. You may use an enrollment count threshold (such as larger than 10) to determine if it is a popular course in the cluster_


````python
def recommend_courses(test_users_labelled, courses_cluster_grouped, threshold=10):
    """
    Recommends popular courses for each test user based on their cluster.

    Parameters:
        test_users_labelled: DataFrame with 'user', 'item', 'cluster'
        courses_cluster_grouped: DataFrame with 'cluster', 'item', 'enrollments'
        threshold: minimum enrollments to consider a course popular

    Returns:
        DataFrame with columns 'user' and 'recommended_courses' (list of items)
    """
    recommendations = []

    # Iterar sobre cada usuario único
    for user_id in test_users_labelled['user'].unique():
        # Subset del usuario
        user_subset = test_users_labelled[test_users_labelled['user'] == user_id]

        # Cursos que ya ha tomado
        enrolled_courses = set(user_subset['item'].tolist())

        # Cluster del usuario
        cluster_id = user_subset['cluster'].iloc[0]

        # Todos los cursos de su cluster con enrollments
        cluster_courses = courses_cluster_grouped[courses_cluster_grouped['cluster'] == cluster_id]

        # Filtrar solo cursos populares según threshold
        popular_courses = set(cluster_courses[cluster_courses['enrollments'] > threshold]['item'].tolist())

        # Recomendaciones = populares pero no tomadas aún
        recommended_courses = list(popular_courses.difference(enrolled_courses))

        # Guardar resultados
        recommendations.append({'user': user_id, 'recommended_courses': recommended_courses})

    return pd.DataFrame(recommendations)

# Aplicar la función
recommendations_df = recommend_courses(test_users_labelled, courses_cluster_grouped, threshold=10)

# Mostrar resultados
recommendations_df.head()
````

````python
# Limitar recomendaciones a máximo 20 cursos
recommendations_df['recommended_courses'] = recommendations_df['recommended_courses'].apply(lambda x: x[:20])

average_recommendations_limited = recommendations_df['recommended_courses'].apply(len).mean()
print(f"Average number of recommended courses per user (max 20): {average_recommendations_limited:.2f}")
````

````python
from collections import Counter

# Aplanar todas las listas de cursos recomendados
all_recommended_courses = [course for sublist in recommendations_df['recommended_courses'] for course in sublist]

# Contar la frecuencia de cada curso
course_counts = Counter(all_recommended_courses)

# Obtener los 10 cursos más recomendados
top_10_courses = course_counts.most_common(10)

# Mostrar resultados
print("Top-10 most frequently recommended courses:")
for course, count in top_10_courses:
    print(f"{course}: {count} times")
````
